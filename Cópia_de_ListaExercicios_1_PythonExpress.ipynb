```python
# 1) Elabore um código que imprima a mensagem *"Olá terráqueos!!"* na tela.

print("Olá terráqueos")
```

```python
numero = input("Digite um número: ")
print(f"O número informado foi {numero}.")
```

```python
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))
soma = num1 + num2
print(f"A soma é: {soma}")
```

```python
nota1 = float(input("Nota 1: "))
nota2 = float(input("Nota 2: "))
nota3 = float(input("Nota 3: "))
nota4 = float(input("Nota 4: "))
media = (nota1 + nota2 + nota3 + nota4) / 4
print(f"A média bimestral é: {media}")
```

```python
metros = float(input("Digite o valor em metros: "))
cm = metros * 100
print(f"{metros}m equivale a {cm}cm.")
```

```python
raio = float(input("Raio do círculo: "))
area = 3.14159 * (raio ** 2)
print(f"A área do círculo é: {area:.2f}")
```

```python
lado = float(input("Lado do quadrado: "))
area_quadrada = lado * lado
dobro = area_quadrada * 2
print(f"A área é {area_quadrada} e o dobro é {dobro}.")
```

```python
valor_hora = float(input("Quanto você ganha por hora? "))
horas_mes = float(input("Quantas horas você trabalhou no mês? "))
salario = valor_hora * horas_mes
print(f"Olá! Este mês seu salário total será de R$ {salario:.2f}. Bom trabalho!")
```

```python
f = float(input("Temperatura em Fahrenheit: "))
c = 5 * ((f - 32) / 9)
print(f"A temperatura em Celsius é: {c:.2f}°C")
```

```python
celsius = float(input("Temperatura em Celsius: "))
fahrenheit = (celsius * 1.8) + 32
print(f"{celsius}°C equivale a {fahrenheit:.2f}°F")
```

```python
num1 = int(input("Digite o primeiro número inteiro: "))
num2 = int(input("Digite o segundo número inteiro: "))
num3 = float(input("Digite um número real (com vírgula): "))

# Cálculos
resultado_a = (2 * num1) * (num2 / 2)
resultado_b = (3 * num1) + num3
resultado_c = num3 ** 3

print(f"a) O produto do dobro do primeiro com metade do segundo: {resultado_a}")
print(f"b) A soma do triplo do primeiro com o terceiro: {resultado_b}")
print(f"c) O terceiro elevado ao cubo: {resultado_c:.2f}")
```

```python
altura = float(input("Digite sua altura (ex: 1.75): "))
peso_ideal = (72.7 * altura) - 58
print(f"Com base na sua altura, seu peso ideal seria: {peso_ideal:.2f} kg")
```
